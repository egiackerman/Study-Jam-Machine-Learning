## 8.2 Memuat dataset Iris

Iris berisi pengukuran **150 bunga** dari tiga jenis: `setosa`, `versicolor`, dan `virginica`. Setiap baris adalah satu contoh bunga.

| Kolom | Arti | Peran |
| --- | --- | --- |
| `sepal length (cm)` | Panjang kelopak bunga dalam cm | Fitur |
| `sepal width (cm)` | Lebar kelopak bunga dalam cm | Fitur |
| `petal length (cm)` | Panjang mahkota bunga dalam cm | Fitur |
| `petal width (cm)` | Lebar mahkota bunga dalam cm | Fitur |
| `species` | Jenis bunga | Label |

**Fitur** adalah informasi input; **label** adalah jawaban yang ingin diprediksi pada supervised learning. Jenis bunga merupakan kategori nominal: tidak ada urutan alami dari setosa ke virginica.

Dataset tersedia bersama scikit-learn, sehingga praktik ini tidak memerlukan unggahan CSV, pemasangan Drive, atau unduhan dataset terpisah.


### Langkah 1 · Memuat data sebagai tabel
Jalankan sel berikut terlebih dahulu.

- `load_iris` adalah fungsi untuk memuat dataset Iris.
- `as_frame=True` meminta representasi tabel pandas.
- `iris.frame.copy()` membuat salinan tabel dan menyimpannya sebagai `df` (*DataFrame*).

Pada tahap ini, tabel masih memiliki kolom `target` dengan kode kelas 0, 1, dan 2.


```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame.copy()
```

### Langkah 2 · Mengubah kode kelas menjadi nama jenis
`iris.target_names` menyimpan nama jenis bunga. `enumerate()` memasangkan urutan angka dengan nama, kemudian `dict()` membentuk pemetaan:

| Kode target | Nama jenis |
| --- | --- |
| 0 | setosa |
| 1 | versicolor |
| 2 | virginica |

`map()` mengganti kode pada `target` dengan nama jenis dan menaruhnya di kolom `species`. Setelah itu, `target` dihapus agar tabel akhir memiliki empat fitur dan satu label. Kode 0, 1, dan 2 menyatakan kategori, bukan tingkat atau besaran.


```python
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))
df = df.drop(columns='target')
```

### Langkah 3 · Memeriksa ukuran dan nama kolom
`df.shape` menghasilkan pasangan `(jumlah_baris, jumlah_kolom)`. `df.columns.tolist()` menampilkan nama kolom sebagai list Python.

**Hasil yang diharapkan:** `(150, 5)`, yaitu 150 contoh bunga dan 5 kolom. Lima kolom tersebut terdiri atas **4 fitur + 1 label**, sehingga bukan berarti ada 5 fitur.


```python
print('Ukuran data:', df.shape)
print('Kolom:', df.columns.tolist())
```

### Langkah 4 · Menampilkan lima baris pertama
`head()` hanya menampilkan lima baris pertama, bukan seluruh dataset. Angka indeks di paling kiri adalah penanda baris dan bukan salah satu dari lima kolom data.

Kelima baris pertama berlabel `setosa`. Itu tidak berarti seluruh dataset hanya berisi setosa; kita akan menghitung semua jenis pada langkah berikutnya.


```python
display(df.head(10))
```

### Langkah 5 · Menghitung contoh untuk setiap jenis
`df['species']` memilih kolom label. `value_counts()` menghitung kemunculan setiap nilai dalam kolom itu.

**Hasil yang diharapkan:** setosa 50, versicolor 50, dan virginica 50. Totalnya 150 bunga. Urutan tampilan kelas bukan hal utama; periksa nama dan jumlahnya.


```python
print('Jumlah per jenis:')
print(df['species'].value_counts())
```

### Membaca satu contoh bunga
Untuk memahami hubungan fitur dan label, lihat satu baris dengan `iloc[0]`. Python memakai indeks mulai dari nol, jadi indeks `0` menunjuk baris pertama.

Empat pengukuran pada baris itu adalah fitur milik satu bunga. `species` adalah label bunga yang sama.


```python
display(df.iloc[0])
```
